```python
def smart_parking(capacity, logs):
    parked = 0
    peak = 0

    for log in logs:
        if log == "IN":
            parked += 1
        elif log == "OUT":
            parked -= 1

        if parked > peak:
            peak = parked

    print("Currently Parked Vehicles:", parked)
    print("Peak Usage:", peak)

    if parked > capacity:
        print("Parking Status: Full - Capacity Exceeded")
    elif parked == capacity:
        print("Parking Status: Full")
    else:
        print("Parking Status: Available")

smart_parking(50, ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"])
```

```python
def estimate_delivery_time(distance, traffic, weather):
    base_time = distance * 5
    if traffic == "High":
        base_time += 15
    elif traffic == "Medium":
        base_time += 8
    if weather == "Rainy":
        base_time += 10
    elif weather == "Storm":
        base_time += 20

    print("Estimated Delivery Time:", base_time, "minutes")
estimate_delivery_time(8, "High", "Rainy")
```

```python
def seat_occupancy(total_seats, booked_seats):
    booked_count = 0

    for seat in booked_seats:
        booked_count += 1

    occupancy = (booked_count / total_seats) * 100
    print("Occupancy:", int(occupancy), "%")

    if occupancy == 100:
        print("Show Status: Housefull")
    elif occupancy >= 75:
        print("Show Status: Almost Full")
        print("Suggestion: Consider Opening Additional Show")
    else:
        print("Show Status: Still Available")
seat_occupancy(200, [1]*150)
```

```python
def server_load(cpu_readings):
    total = 0

    for reading in cpu_readings:
        total += reading

    average = total / len(cpu_readings)
    print("Average CPU Load:", int(average), "%")

    if average < 50:
        print("Server Status: Normal")
    elif 50 <= average <= 80:
        print("Server Status: Warning")
    else:
        print("Server Status: Critical")
server_load([45, 60, 70, 85, 90])
```

```python
def resource_monitor(resources):
    overused = []

    for resource, hours in resources.items():
        if hours > 8:   # Overuse condition
            overused.append(resource)

    print("Overused Resources:", ", ".join(overused) if overused else "None")

    if overused:
        print("Energy Alert: Yes")
    else:
        print("Energy Alert: No")
usage = {
    "Projector": 6,
    "AC": 9,
    "Lights": 4
}

resource_monitor(usage)
```

```python
def event_registration(capacity, registrations):
    if registrations > capacity:
        confirmed = capacity
        waitlisted = registrations - capacity
        status = "Closed"
    else:
        confirmed = registrations
        waitlisted = 0
        status = "Open"

    print("Confirmed Registrations:", confirmed)
    print("Waitlisted Users:", waitlisted)
    print("Registration Status:", status)
event_registration(100, 105)
```
