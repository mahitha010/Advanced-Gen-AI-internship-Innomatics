```python
marks = [45, 78, 90, 33, 60]
for result in marks:
  if result >= 50:
    print("pass")
  else:
    print("fail")
print(f"Total Students: {len(marks)}")
passed_students = 0
for result in marks:
  if result >= 50:
    passed_students += 1
    print("pass")
print(f"Passed Students: {passed_students}")
failed_students = len(marks) - passed_students
print(f"Failed Students: {failed_students}")
total_students = len(marks)
print(f"Total Students: {total_students}")
```

```python
names = [" Alice ", "bob", " CHARLIE "]
names=[item.lower() for item in names]
print(names)
cleaned_list=names.copy()
for item in names:
    cleaned_list.remove(item)
print(cleaned_list)
```
