Task 1: Conceptual Understanding
1. Difference between "Love" and "love" in NLP

In NLP, "Love" and "love" are treated as different tokens if case normalization is not applied because machines treat text as exact strings.

Example:

Love → Token 1
love → Token 2

This creates data inconsistency and increases vocabulary size unnecessarily.

After lowercasing:

Love → love

Both become same token which improves model learning.

2. What happens if stopwords are not removed?

If stopwords are not removed:

• Vocabulary size increases
• Model training becomes slower
• Noise increases in dataset
• Important words get less importance
• Memory usage increases

Example:

Sentence:
"I am going to the store"

Without stopword removal:
["I","am","going","to","the","store"]

With removal:
["going","store"]

3. Two real-world cases where removing stopwords is harmful
Case 1: Sentiment Analysis

Sentence:

"I am not happy"

If stopwords removed:

["happy"]

Meaning becomes opposite.

Case 2: Question Answering Systems

Sentence:

"What is the time?"

Removing stopwords:

["time"]

Intent is lost.

4. Difference between stemming and lemmatization
Stemming

Cuts words using rules without understanding meaning.

Example:

playing → play
studies → studi
better → better

Fast but sometimes incorrect.

Lemmatization

Uses vocabulary and grammar to convert words to root form.

Example:

playing → play
studies → study
better → good

Slower but more accurate.

```python
#Task 2: Advanced Preprocessing Function

import re
import string
from collections import Counter
```

```python
def preprocess_text(text):

    if not isinstance(text,str):
        return [],""

    if text.strip()=="":
        return [],""
    text=text.lower()
    text=re.sub(r'http\S+|www\S+', '', text)
    text=re.sub(r'\S+@\S+', '', text)
    text=re.sub(r'\d+','',text)
    text=re.sub(r'[^\w\s]', ' ', text)
    text=re.sub(r'(.)\1+', r'\1', text)
    text=re.sub(r'\s+',' ',text).strip()

    tokens=text.split()

    clean_tokens=[]

    for word in tokens:

        if len(word)<=2 and word not in ['no','not']:
            continue

        clean_tokens.append(word)

    clean_sentence=" ".join(clean_tokens)

    return clean_tokens,clean_sentence
```

```python
#Task 3: Stress Testing

sentences=[

"Get 100% FREE access now!!!",
"I absolutely looooved this product 😍😍",
"Worst service ever... 0/10",
"Call me at 9876543210",
"This is THE best course!!!",
"Visit https://openai.com now!",
"Nooooo this is baaad!!!",
"OK OK OK I got it",
"Win $$$ now!!! Limited offer!!!",
"I am not happy with this"

]
```

```python
results=[]

for s in sentences:

    tokens,clean_sentence=preprocess_text(s)

    print("Original:",s)
    print("Tokens:",tokens)
    print("Cleaned:",clean_sentence)
    print()

    results.append(tokens)
```

```python
#Task 4: Token Analytics

for i,tokens in enumerate(results):

    if len(tokens)==0:
        continue

    total=len(tokens)

    unique=len(set(tokens))

    avg=sum(len(word) for word in tokens)/total

    print("Sentence",i+1)

    print("Total tokens:",total)

    print("Unique tokens:",unique)

    print("Average length:",round(avg,2))

    print()
```

```python
#task 5

all_tokens=[]

for t in results:
    all_tokens.extend(t)

freq=Counter(all_tokens)

print(freq.most_common(10))
print(freq.most_common()[-5:])
```

```python
#task 6

def full_pipeline(text_list):

    all_tokens=[]
    clean_sentences=[]

    for text in text_list:

        tokens,clean=preprocess_text(text)

        all_tokens.extend(tokens)

        clean_sentences.append(clean)

    return {

        "tokens":all_tokens,

        "clean_sentences":clean_sentences

    }
    output=full_pipeline(sentences)

    print(output)
```

```python
#Task 7: Error Handling

edge_cases=[

"",
"😍😍😍😍",
"123456",
"     "

]

for case in edge_cases:

    tokens,clean=preprocess_text(case)

    print("Input:",case)

    print(tokens)

    print(clean)

    print()
```
